# đọc và hiển thị ảnh

# Đổi màu

```python
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)

lab = cv2.cvtColor(img, cv2.COLOR_BGR2LAB)

plt.figure(figsize=(15, 5))

plt.subplot(131)
plt.imshow(gray, cmap='gray')
plt.title('Gray')

plt.subplot(132)
plt.imshow(hsv)
plt.title('HSV')

plt.subplot(133)
plt.imshow(lab)
plt.title('LAB')

plt.show()
```

# Cắt ảnh

```python
#cắt theo kích thước cụ thể (y1:y2 , x1,x2)
crop_img = img[600:1000, 500:1000]

crop_img_gray = cv2.cvtColor(crop_img, cv2.COLOR_BGR2GRAY)
plt.imshow(crop_img_gray, cmap='gray')
plt.axis("off")
plt.show
```

```python
# Thay đổi kích thước (Resize) về 300x300 pixel
resize_img = cv2.resize(img, (300, 300))

plt.imshow(resize_img)
plt.axis("off")
plt.show()
```

```python
#Resize theo tỉ lệ 
resized_half = cv2.resize(img, (0,0), fx=0.5, fy=0.5)
plt.imshow(resize_img)
plt.axis("on")
plt.show()
```

# Vẽ hình cơ bản và thêm văn bản

```python
points = [
    (300, 100),  
    (350, 250),
    (500, 250),
    (375, 350),
    (425, 500),
    (300, 400),
    (175, 500),
    (225, 350),
    (100, 250),
    (250, 250)
]

# Vẽ các cạnh
for i in range(len(points)):
    lineimg = cv2.line(img, points[i], points[(i+1) % len(points)], (0, 255, 255), 2)
# Vẽ hình chữ nhật
cv2.rectangle(img, (50, 50), (200, 200), (0, 255, 0), 3)
# Vẽ hình tròn (tâm, bán kính, màu, độ dày)
cv2.circle(img, (300, 300), 50, (0, 0, 255), -1) # -1 là tô kín
cv2.putText(img, 'TIRED', (400, 850), cv2.FONT_HERSHEY_SIMPLEX, 10, (255, 255, 255), 2)
plt.imshow(lineimg)
plt.axis("off")
plt.show()
```
